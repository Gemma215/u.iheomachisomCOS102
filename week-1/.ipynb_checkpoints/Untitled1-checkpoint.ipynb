```python
# Integer number
num = 100
print(num)
print("Data Type of variable num is", type(num))

# float number
fnum = 34.45
print(fnum)
print("Data Type of fnum is", type(fnum))

#complex number
cnum = 3 + 4j
print(num)
print("Data Type of variable cnum is", type(cnum))
```

# Financial project

```python
# Simple Interest
P = float(input("Enter Principal: "))
R = float(input("Enter Rate: "))
T = float(input("Enter Time: "))

A = P * (1 + (R / 100) * T)

print(f"The Simple Interest Amount (A) is: {A:.2f}")
```

```python
# Compound Interest
n = int(input("Times compounded per year: "))
t = float(input("Number of years: "))

A = P * (1 + R/n)**(n*t)

print(f"The Compound Interest  (A) is: {A:.2f}")
```

```python
# Annuity Plan
PMT = float(input("Enter periodic payment:"))
numerator = ((1 + R/n)**(n*t)) - 1
denominator = R/n

A = PMT * (numerator / denominator)

print(f"The Annuity Plan Future Value (A) is: {A:.2f}")
```
