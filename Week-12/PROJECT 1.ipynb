```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.read_csv('titanic.csv')


print(" TITANIC DATASET OVERVIEW")
print(f"Total Passengers on board: {len(df)}")
print(f"Average Ticket Fare Paid: ${df['Fare'].mean():.2f}")
print(f"Oldest Passenger: {df['Age'].max()} years old")
print(f"Youngest Passenger: {df['Age'].min()} years old")
print("-" * 40)

print("\n--- First 10 Passengers ---")
df.head(10)
```

```python
print(" SURVIVAL RATE ANALYSIS ")


print("\n1. Survival Rate by Gender:")
gender_survival = df.groupby('Sex')['Survived'].mean() * 100
for sex, rate in gender_survival.items():
    print(f"   • {sex.title()}: {rate:.1f}% survived")


print("\n2. Survival Rate by Passenger Class:")
class_survival = df.groupby('Pclass')['Survived'].mean() * 100
class_names = {1: "1st Class (Rich)", 2: "2nd Class (Middle)", 3: "3rd Class (Economy)"}
for pclass, rate in class_survival.items():
    print(f"   • {class_names[pclass]}: {rate:.1f}% survived")
```

```python

plt.figure(figsize=(7, 5))
sns.countplot(data=df, x='Survived', palette=['#ff6b6b', '#4cd137'])


plt.xticks([0, 1], ['Did Not Survive (0)', 'Survived (1)'])
plt.title('Count of Survivors vs Non-Survivors', fontsize=14, fontweight='bold')
plt.xlabel('Status', fontsize=12)
plt.ylabel('Number of Passengers', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.5)


plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.histplot(data=df, x='Age', bins=20, kde=True, color='#4a69bd')


plt.title('Distribution of Passenger Ages on the Titanic', fontsize=14, fontweight='bold')
plt.xlabel('Age (Years)', fontsize=12)
plt.ylabel('Passenger Count', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.5)


plt.show()
```
