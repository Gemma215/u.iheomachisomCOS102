```python
import tkinter as tk
from tkinter import messagebox
import pandas as pd
import os


CSV_FILE = "GIG-logistics.csv"

def verify_employee():
    
    input_name = entry_name.get().strip()
    input_dept = entry_dept.get().strip()
    
    
    if not input_name or not input_dept:
        messagebox.showwarning("Input Error", "Please fill in both Name and Department fields.")
        return

    
    if not os.path.exists(CSV_FILE):
        messagebox.showerror("File Error", f"Database file '{CSV_FILE}' not found!\nPlease ensure it is in the same directory.")
        return

    try:
    
        df = pd.read_csv(CSV_FILE)
        
        
        df['Name'] = df['Name'].astype(str).str.strip()
        df['Department'] = df['Department'].astype(str).str.strip()
        
    except Exception as e:
        messagebox.showerror("Error", f"Could not read the CSV file.\nDetails: {e}")
        return

    
    matched_employee = df[
        (df['Name'].str.lower() == input_name.lower()) & 
        (df['Department'].str.lower() == input_dept.lower())
    ]

    if not matched_employee.empty:
        
        text_results.delete("1.0", tk.END)
        
    
        official_name = matched_employee.iloc[0]['Name']
        official_dept = matched_employee.iloc[0]['Department']
        
        
        welcome_msg = f"Welcome, {official_name}!\n"
        welcome_msg += f"Verification successful for the {official_dept} department.\n"
        welcome_msg += "-" * 40 + "\n"
        welcome_msg += f"Other members in the {official_dept} department:\n\n"
        
        text_results.insert(tk.END, welcome_msg)
        
    
        dept_members = df[df['Department'].str.lower() == official_dept.lower()]
        other_members = dept_members[dept_members['Name'].str.lower() != official_name.lower()]
        
        if not other_members.empty:
            for idx, row in other_members.iterrows():
                text_results.insert(tk.END, f"• {row['Name']}\n")
        else:
            text_results.insert(tk.END, "No other employees found in this department.")
            
    else:
        messagebox.showinfo("Access Denied", f"We are sorry, but '{input_name}' under the '{input_dept}' department does not exist in our employee database.")
        text_results.delete("1.0", tk.END)



root = tk.Tk()
root.title("GIG Logistics - Employee Verification System")
root.geometry("500 leaping" if False else "500x500")
root.resizable(False, False)
header_label = tk.Label(root, text="GIG LOGISTICS", font=("Arial", 16, "bold"), fg="#1c3d5a")
header_label.pack(pady=15)

subheader_label = tk.Label(root, text="Employee Identity & Attendance Portal", font=("Arial", 10, "italic"))
subheader_label.pack(pady=2)


form_frame = tk.Frame(root)
form_frame.pack(pady=15)

lbl_name = tk.Label(form_frame, text="Full Name:", font=("Arial", 11))
lbl_name.grid(row=0, column=0, sticky="w", padx=5, pady=5)
entry_name = tk.Entry(form_frame, font=("Arial", 11), width=30)
entry_name.grid(row=0, column=1, padx=5, pady=5)


lbl_dept = tk.Label(form_frame, text="Department:", font=("Arial", 11))
lbl_dept.grid(row=1, column=0, sticky="w", padx=5, pady=5)
entry_dept = tk.Entry(form_frame, font=("Arial", 11), width=30)
entry_dept.grid(row=1, column=1, padx=5, pady=5)


btn_verify = tk.Button(root, text="Verify Identity", font=("Arial", 11, "bold"), bg="#1c3d5a", fg="white", width=20, command=verify_employee)
btn_verify.pack(pady=10)


result_frame = tk.Frame(root)
result_frame.pack(pady=10, fill="both", expand=True, padx=20)

lbl_results = tk.Label(result_frame, text="Results / Department Directory:", font=("Arial", 10, "bold"))
lbl_results.pack(anchor="w")


text_results = tk.Text(result_frame, font=("Arial", 10), height=12, wrap="word")
text_results.pack(fill="both", expand=True, pady=5)


root.mainloop()
```
