```python

class Item:
    def __init__(self, name, price):
        self.name = name
        self.price =
    def calculate_total(self, quantity):
        total = self.price * quantity
        return total

    def display_info(self):
        return f"Product: {self.name} | Original Price: ${self.price:.2f}"



class DiscountedItem(Item):
    def __init__(self, name, price, discount_percentage):
        super().__init__(name, price)
        self.discount_percentage = discount_percentage

    
    def calculate_total(self, quantity):
        original_total = super().calculate_total(quantity)
        discount_amount = original_total * (self.discount_percentage / 100)
        final_total = original_total - discount_amount
        return final_total

    
    def display_info(self):
        return f"Product: {self.name} | Original Price: ${self.price:.2f} ({self.discount_percentage}% OFF)"



print("=== OOP Inheritance & Polymorphism Demo ===")


standard_item = Item("Python Coding Textbook", 50.00)
sale_item = DiscountedItem("Gaming Laptop", 1200.00, 15)


shopping_cart = [standard_item, sale_item]


print("\n🛒 Processing Shopping Cart Items:")
for product in shopping_cart:
    print("-" * 50)
    print(product.display_info())
    
    quantity = 2
    cost = product.calculate_total(quantity)
    print(f"Cost for {quantity} units: ${cost:.2f}")

print("-" * 50)
print("\n Code execution complete!")
```
