```python
def calculate_atr():
    print("--- Izifin Technology: Staff ATR Calculator ---")
    
    try:
        # Taking inputs
        experience = float(input("Enter years of work experience: "))
        age = float(input("Enter age of the staff: "))

        # Logic hierarchy (Checking most senior first)
        if experience > 25 and age >= 55:
            atr = 5600000
        elif experience > 20 and age >= 45:
            atr = 4480000
        elif experience > 10 and age >= 35:
            atr = 1500000
        else:
            # Default category for staff with less experience/age
            atr = 550000

        # Display result formatted with commas for readability
        print(f"\nThe Annual Tax Revenue (ATR) for this staff is: ₦{atr:,}")

    except ValueError:
        print("Invalid input. Please enter numerical values for age and experience.")

if __name__ == "__main__":
    calculate_atr()
```
