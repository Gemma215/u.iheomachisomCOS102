# TASK A

```python
BEGIN
    
    SET subtotal = 0
    SET tax_rate = 0.08  (8% tax)
    SET more_items = TRUE

    
    WHILE more_items IS TRUE
        INPUT item_price
        subtotal = subtotal + item_price
        
        INPUT "Is there another item? (Yes/No)"
        IF response IS "No" THEN
            SET more_items = FALSE
        END IF
    END WHILE

    
    total_tax = subtotal * tax_rate
    final_bill = subtotal + total_tax

    DISPLAY "Subtotal: ", subtotal
    DISPLAY "Tax: ", total_tax
    DISPLAY "Total Amount Due: ", final_bill

    // Step 4: Handle Payment
    INPUT payment_received
    
    IF payment_received >= final_bill THEN
        change = payment_received - final_bill
        DISPLAY "Transaction Successful!"
        DISPLAY "Your change is: ", change
    ELSE
        DISPLAY "Error: Insufficient funds. Please provide more cash."
    END IF

END
```
