```python
import pandas as pd
import numpy as np

print("Step 1: Loading the Titanic dataset...")
try:
    
    df = pd.read_csv('titanic.csv')
    print("-> Success! Titanic file loaded perfectly.")
    
    print("\nStep 2: Cleaning missing data...")

    median_age = df['Age'].median()
    df['Age'] = df['Age'].fillna(median_age)
    
    
    df['Cabin'] = df['Cabin'].fillna("Unknown")
    
    
    most_common_port = df['Embarked'].mode()[0]
    df['Embarked'] = df['Embarked'].fillna(most_common_port)
    print("-> Success! Missing ages, cabins, and ports have been handled.")

    print("\nStep 3: Fixing text formatting and removing duplicates...")
    
    text_cols = ['Name', 'Sex', 'Ticket', 'Cabin', 'Embarked']
    for col in text_cols:
        df[col] = df[col].astype(str).str.strip()
        
    
    df.drop_duplicates(inplace=True)
    print("-> Success! Text formatting is clean and duplicates removed.")

    print("\nStep 4: Saving your brand new clean Titanic file...")
    
    df.to_csv('titanic_cleaned.csv', index=False)
    print("-> SUCCESS! Your clean file is saved as 'titanic_cleaned.csv'")
    
    
    print("\n--- Here is a sneak peek of your cleaned Titanic data: ---")
    print(df[['Name', 'Sex', 'Age', 'Cabin', 'Embarked']].head())

except FileNotFoundError:
    print("\n ERROR: Python can't find 'titanic.csv'.")
    print("Please go to your Jupyter dashboard and make sure 'titanic.csv' is in your Week-4 folder!")
except PermissionError:
    print("\n ERROR: Permission Denied.")
    print("Please close Microsoft Excel if you have the 'titanic.csv' file open inside it, then retry!")
except Exception as e:
    print(f"\n An unexpected error occurred: {e}")
```
